```python

from pathlib import Path
import subprocess
import sys

# El notebook debe estar dentro de la carpeta "notebooks".
ROOT = Path.cwd()

if ROOT.name.lower() == "notebooks":
    ROOT = ROOT.parent

APP_PATH = ROOT / "app.py"

if not APP_PATH.is_file():
    raise FileNotFoundError(
        f"No se encontró app.py en:\n{APP_PATH}\n"
        "Coloca app.py en la raíz de Predict_Solubility_v1.3."
    )

# Utiliza el mismo entorno Conda que ejecuta el notebook.
PYTHONW = Path(sys.executable).with_name("pythonw.exe")

if not PYTHONW.is_file():
    raise FileNotFoundError(
        f"No se encontró pythonw.exe en:\n{PYTHONW}\n"
        "Comprueba que estás usando el entorno sci_env de Windows."
    )

# Abre la interfaz en un proceso independiente de Jupyter.
subprocess.Popen(
    [str(PYTHONW), str(APP_PATH)],
    cwd=str(ROOT)
)

print("Aplicación lanzada en un proceso independiente de Jupyter.")
```
